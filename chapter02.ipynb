### 논리연산자

- and(둘 다 참)
- or (둘 중 하나)
- not(부정)

```python
is_adult = True
has_ticket = False

#and 연산
print(is_adult and has_ticket) # False

#or 연산
print(is_adult or has_ticket) # True

#not 연산
print(not is_adult) # False
```

```python
print(False and (10/0 > 1))
print(False or (10/0 > 1))
```

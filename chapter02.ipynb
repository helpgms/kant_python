### 논리연산자

- and(둘 다 참)
- or (둘 중 하나)
- not(부정)

````python
is_adult = True
has_ticket = False

#and 연산
print(is_adult and has_ticket) # False

#or 연산
print(is_adult or has_ticket) # True

#not 연산
print(not is_adult) # False
````

### 문자열


````python
print("선발 출전: %s, %d" % ("손흥민",7))
print("선발 출전: [%5d]" % (7))
print("소숫점 표기: %.2f" % 3.141592)
````

````python
name = "Cake"
price = 6500
qty = 7

# format 방식
format_str = "주문내역: {0} {1} -> 총 가격 : {2}원" .format(name,qty,price * qty)
print("상품명 {}, 가격 {} 원, 수량 {}개 ".format(name,price,qty))
print(format_str)

#f-string 방식
format_str = f"주문내역: {name} {qty} -> 총 가격 : {(price * qty):>10,}원"
print(format_str)
````

````python
print(f"금액: {15000:>8,}원")
````

````python
# 가상의 주문 데이터
orders = [
    ("Latte", 5000, 2),
    ("Americano", 3000, 1),
    ("Cake", 6000, 1)
]

print("=" * 33)
print(f"{'메뉴명':<10} {'단가':>8}{'금액':>9}")
print('-' * 33)

# 총금액
total_amount = 0

for menu, price, qty in orders:
    subtotal = price * qty
    total_amount += subtotal
    print(f"{menu:<10}{price:>13}{subtotal:>10,}")

print("-"*33)
print(f"{'합계금액:':<10}{total_amount:>19,}원")

````

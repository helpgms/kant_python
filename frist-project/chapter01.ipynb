### 변수 이름 규칙

- 숫자로 시작 불가 : 1_name (x)
- 공백 불가 : user name (x)
- 하이픈(-) : user-age (x)
- 예약어 사용불가 : if, for, def (x)

### 컨벤션

- 스네이크 표기법 : user_name, value_list

### 다른 표기법

- 파스칼 표기법 : UserName, ValueList
- 카멜 표기법(단봉 낙타법) : userName, valueList 

````python
# 정수형 자료형
age = 25 
# 실수형 자료형
pi = 3.14 
# 불리언 자료형(Boolean)
is_alive = True
# 문자열 자료형
user_name = "KangMinSeong"
# 배열 Array
grade_type = [1, 2, 3, 4]
````

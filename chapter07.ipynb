### Function 함수

- 단독적으로 사용 가능 ex. print(), len(), ....
- 매개변수(Parameter)
- 반복되는 로직을 합쳐놓은 문법
- 재사용 높일 수 있는 코드 기법
- 함수명 컨벤션 : snack_case 사용

```python
def 함수이름 (매개변수, 매개변수, ....):
    실행할 로직 #1
    실행할 로직 #2
    실행할 로직 #3
```

````python
# 함수 정의
# default 값은 맨앞에 올 수 없다. 맨 뒤에 있어야한다. 기본값이 2개일때는 그 자체가 맨뒤에 있어야한다. ex) (name, age = 20, point = 1000)

def say_hello(name, age, point = 1000):
    print(f"안녕하세요. 저는 {name}입니다. 나이는 {age}입니다")
    print(f"초기 지급 포인트 {point}점이 지급되었습니다.")

say_hello("kang" , 2, 1)
````

````python
# 가변 인수
# *args 이렇게 많이 씀

def add_all(*numbers): # 매개변수(Parameter)
    print(f"입력된 튜플: {numbers}")
    total = sum(numbers)
    print(f"합계 결과: {total}")

add_all(1, 2, 3) # 인수(argument)
add_all(1, 2, 3, 4, 5)
````

````python
# 딕셔너리 가변 인수
# **kwargs

def user_info(**data):
    print(f"전달된 딕셔너리 구조: {data}")
    for key, value in data.items():
        print(f"{key} : {value}")

# 함수 호출

user_info(name = "sparta" , level = 5, active = True)
````

````python
# 반환값 return - 결과값 되돌리기
# 반환값 1개

def check_ord_number(number):
    if number > 0:
        return "양수"
    else:
        return "음수"

print(check_ord_number(10))
print(check_ord_number(-20))
````

````python
# 여러개 결과값 반환

def calc_value(a,b):
    plus = a + b
    minus = a - b
    return plus, minus # 자동으로 튜플로 리턴

result1, result2 = calc_value(10,5)
print(f"덧셈: {result1}, 뺄셈: {result2}")

res = calc_value(10,5)
print(f"튜플: {res}")
type(res)
````

## 변수 스코프 (유효 범위)

- 전역 변수 : 어디서나 접근 가능, 함수 밖에서 선언하는 변수
- 지역 변수 : 함수 내에서 선언하는 변수, 함수 안에서만 접근 가능

````python
def local():
    local_var = 1 # 지역변수
    print(local_var)

local()
# print(local_var) # 오류 발생
````

````python
# 전역변수
message = "안녕하세요. kant몰에 오신 것을 환영합니다"

def greeting():
    message = "장난감 코너는 3층에 있습니다"
    print(message) # 지역변수가 우선된다

greeting()

print(message) # 전역변수
````

````python
# 전역 변수
value = "전역 데이터"

# Function==================
def outer_function():
    value = "지역 변수1"

    def inner_function():
        value = "지역 변수2"
        print(value)
    
    inner_function()
    print(value)
# Function==================

outer_function()
print(value) # 전역변수
````

````python
# global 키워드

counter = 0

def inc_counter() :
    global counter # 전역 변수를 참조하겠다.
    counter += 1

inc_counter()
inc_counter()

print(counter)
````

## Lambda Expression 람다식 (함수라 생각)

```python
lambda 매개변수: 표현식(로직)

```

````python
# 일반 함수
def add_five(x):
    return x + 5

print(add_five(10))

# 람다식
res = lambda x : x + 5
print(res(10))
````

````python
# map 모든 요소를 반환하는 기능
numbers = [1, 2, 3, 4, 5]
res = list(map(lambda x : x + 10, numbers))
print(f"계산 결과 : {res}")

# 리스트 컴프리핸션
print([x+10 for x in numbers])

# filter 조건에 맞는 요소 반환하는 기능
filter_res = list(filter(lambda x : x % 2 == 0, numbers))
print(f"짝수 결과: {filter_res}")

print([x for x in numbers if x % 2 == 0])
````

````python
# 정렬
student = [
    ("손", 95), 
    ("이", 60), 
    ("강", 80)
]

print(sorted(student, key=lambda s : s[1])) # 오름
print(sorted(student, key=lambda s : s[1], reverse = True)) # 내림
print(max(student, key=lambda s : s[1]))
````

````python
numbers = [2, 3, 4, 5, 6, 7]

# Lambda 식으로 모든 요소의 거듭제곱 추출

sqr_numbers = list(map(lambda n : n ** 2, numbers))
print(f"거듭제곱 결과값: {sqr_numbers}")
print([n ** 2 for n in numbers])

# 결과값이 20이상인 값만 추출
res_numbers = list(filter(lambda s : s >= 20, sqr_numbers))
print(f"결과값이 20이상인 값 : {res_numbers}")
# print([n for n in numbers if n >= 20])
````

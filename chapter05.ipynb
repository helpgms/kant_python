````python
# 다양한 조건식

cart = {"카메라" : 50000, "삼각대" : 70000}

if cart:
    total_amount = sum(cart.values())
    print(f"총 금액 :{total_amount},  결제를 진행합니다.")
else:
    print("장바구니가 비었습니다")
````

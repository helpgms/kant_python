### FOR 반복문

```python
for 변수 in 반복할_데이터:
   반복수행할 코드 #1
   반복수행할 코드 #2
```

````python
# 리스트
fruits = ["사과", "바나나", "체리"]

for fruit in fruits:
    print(fruit)
````

````python
for ch in "스파르타":
    print(ch)
````

````python
menu = {"coffee" : 2000, "tea" : 2500, "latte" : 4000}

menu.keys()

for key in menu:
    print(key)

for name, price in menu.items():
    print(f"{name} : {price}")
````

````python
# range()

for i in range(5):
    print(f"{i+1} 번째 로직 수행")
````

````python
# 범위 지정

for i in range(1, 6, 2):
    print(i)

print(range(5))
print(list(range(5))) # 리스트로 반환 후 출력
````

### while 반복문

````python
num = 1

while num <=3:
    print(f"{num}번째입니다")
    num += 1
````

````python
count = 3

while count > 0:
    print(count)
    count -= 1
````

````python
for num in range(1,6):
    if num == 3:
        print("즉시 종료")
        break
    print(f"현재 번호 = {num}")
````

````python
while True:
    user_input = input()
    if user_input == "q":
        print("종료")
        break
````

````python
for num in range(1,6):
    if num % 2 == 0:
        continue
    print(f"난 홀수{num} 다")
````

````python
# 구구단
for i in range(2,10,1):
    if i == 4:
        break
    print(f"---{i}단---")
    for o in range(1,10,1):
        if o == 6:
            break
        print(f"{i} x {o} = {i*o}")
````

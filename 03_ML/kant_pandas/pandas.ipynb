```python
amounts = [3200, 1500, 15000, 6500, 32000, 4800,
           10900, 1500, 2700, 3000, 18500, 7000]

print(amounts)
print("합계", sum(amounts))

print(amounts * 2)
```

```python
import numpy as np

arr  = np.array(amounts)
arr * 2

print("합계 : ", arr.sum())
print("평균 : ", round(arr.mean()))
print("최소 : ", arr.min())
print("최대 : ", arr.max())

print("모양 :", arr.shape, " 데이터 타입 : ", arr.dtype)
```

```python
print(arr >= 10000)
arr[arr >= 10000]  # [조건식]에 맞는 애들만 추출
```

```python
amounts = [3200, 1500, 15000, 6500, 32000, 4800,
           10900, 1500, 2700, 3000, 18500, 7000]

shops = ["편의점", "버스", "영화관", "분식집", "온라인쇼핑", "카페",
         "음원앱", "버스", "편의점", "웹툰", "온라인쇼핑", "분식집"]

categories = ["식비", "교통", "문화", "식비", "쇼핑", "식비",
              "문화", "교통", "식비", "문화", "쇼핑", "식비"]

print(shops)

food = 0
for c, a in zip(categories, amounts):
    if c == "식비":
        food = food + a

print(food)
```

```python
import pandas as pd

spend = pd.DataFrame({
    "day"  : [2, 3, 4, 5, 7, 9, 12, 14, 16, 19, 20, 25 ], 
    "shop" : shops,
    "categories" : categories,
    "amounts" : amounts
})

spend.head(5)
spend.info()
# print(spend.shape)
```

```python
spend["amounts"]
print("합계 : ", spend["amounts"].sum())
print("평균 : ", round(spend["amounts"].mean(),2))
```

```python
# spend["shop", "amounts"]
# spend[ 컬럼명 ]
# spend[[컬럼1, 컬럼2]]
spend[["shop", "amounts"]]
```

### 조건 설정

```python
# 조건식
spend["categories"] == "문화"
```

```python
# spend[ 조건 ] : 조건이 TRUE인 행만 리턴
# spend[ "컬럼명" ] : "컬럼1"의 로우(row) 모두 리턴
spend[spend["categories"] == "문화"]
```

```python
# 조건식
spend[spend["amounts"] >= 10000]
```

```python
# 식비 5000원 이상 사용한 행을 출력
# spend["categories"] == "식비" & spend["amounts"] >= 5000
# 연산자 우선순위에서 "" 이게 더 높아서 &가 뒤늦게 실행이 됨
# 따라서 () 우선순위를 정해주고 & 해서 확실하게
spend[(spend["categories"] == "식비") & (spend["amounts"] >= 5000)]
```

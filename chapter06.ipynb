```python
# 회원 데이터
user_database = [
    {"name" : "황인범", "level" : 'VIP', "active" : True, "point" : 15000},
    {"name" : "김민재", "level" : 'Gold', "active" : False, "point" : 13000},
    {"name" : "조규성", "level" : 'VIP', "active" : True, "point" : 1500},
    {"name" : "손흥민", "level" : 'Gold', "active" : True, "point" : 32000},
    {"name" : "이강인", "level" : 'Sliver', "active" : False, "point" : 4000}
]

# VIP 포인트 누적 변수
total_vip_points = 0
#현재 활성 회원 수
active_user_count = 0 

print("=== 마케팅 대상 활성 VIP 및 Gold 회원 결과")

for user in user_database:
    # 활성화 된 고객 필터링 active
    if user["active"]:
        # 활성 회원 수 누적
        active_user_count += 1

        if user["level"] == "VIP":
            total_vip_points += user["point"]
            print(f"[VIP 등급] {user['name']} 회원님에게 VIP쿠폰이 발행되었습니다")
        elif user["level"] == "Gold":
            print(f"[Gold 등급] {user['name']} 회원님에게 Gold쿠폰이 발행되었습니다")

print(f"전체 활성 회원 수 : {active_user_count}명")
print(f"활성 VIP 포인트 누적 : {total_vip_points}점")
```

```python
import random # 난수 발생을 위한 모듈 임포트

# 1~100 사이 난수 발생
num = random.randint(1,100)
try_count = 0

# 타이틀
print("=" * 30)
print("난수 발생 완료")
print("=" * 30)

# 메인 루프
while True:
    user_input = input("입력하세요 : ")

    if user_input == "q":
        print(f"아쉽군요, 정답은 {num}입니다")
        break
    
    if not user_input.isdecimal(): # 숫자가 아니면 다시
        print("숫자만 입력해주세요.")
        continue

    user_number = int(user_input)

    if user_number < 0 or user_number > 100:
        print("다시 입력해주세요")
        continue

    # 횟수 증가
    try_count += 1
    # 입력값과 비교
    if user_number == num:
        print("정답입니다.")
        print(f"시도한 횟수는 {try_count}회 입니다")
        break
    elif user_number > num:
        print("입력하신 수보다 작아요")
    elif user_number < num:
        print("입력하신 수보다 커요")
```

## 리스트와 튜플

### 리스트
수정 가능
- 대괄호 [ ]

### 튜플
수정 불가능, 고정
- 소괄호 ( )

### 딕셔너리 
키와 벨류를 가지고 있음
- 중괄호 { }

````python
# 리스트 : 가변 데이터 저장
shopping_list =["apple", "banana", "cherry"]

print(shopping_list)
print(shopping_list[0])

shopping_list[0] = "kiwi"
print(shopping_list[0])
print(shopping_list)
````

---
### 리스트 메서드

````python
fruits = ["apple"]

#append(뒤에 추가)
fruits.append("banana")
print(fruits)

#insert(번호에 추가)
fruits.insert(0,"kiwi")
print(fruits)

#remove(삭제)
fruits.remove("apple")
print(fruits)

fruits.append("grape")
print(fruits)

#  멀티 주석하는 방법
# -> ctrl + /
````

### 튜플(Tuple)
- 한번 생성하면 요소를 추가, 수정, 삭제 할 수 없다.

````python
axis = (38.123, 41.123)
# axis[0] = 36.1234

point = 1, 2, 3
print(point,type(point))
````

````python
import sys
print(sys.getsizeof([1,2,3])) 
print(sys.getsizeof((1,2,3))) # 튜플: 데이터 안정성
````

### 슬라이싱
- 정렬할때 이용

````python
numbers = [0, 1, 2, 3, 4, 5]

# 1번부터 4번 미만까지 슬라이싱
numbers[1:4]

# 2칸 간격으로 추출
numbers[::2]

# 역순 추출
numbers[::-1]

# 앞에서 3개만 추출
number[:3]

numbers[3:]
````

````python
print((10, 20, 30, 40)[1:3])
````

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

print(matrix[1])
print(matrix[1][2])
print(len(matrix), len(matrix[0]))
````
